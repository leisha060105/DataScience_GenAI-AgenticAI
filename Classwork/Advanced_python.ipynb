LOOPS

```python
i = 1  #initialize the counter

while i <= 5:  #condition
    print('data science')
    i += 1  #increment the counter
```

```python
i = 1  #initialize the counter

while i <= 5:  #condition
    print('data science :', i)
    i += 1  #increment the counter
```

```python
i = 5  #initialize the counter

while i >= 1:  #condition
    print('data science :', i)
    i -= 1  #decrement the counter
```

```python
i = 1

while i <= 5:  #condition
    print('data science :', i)
    j = 1
    while j <= 4:
        print('technology')
        j += 1

    i += 1  #increment the counter
    print()
```

```python
i = 1

while i <= 5:  #condition
    print('data science :', end = ' ')
    j = 1
    while j <= 4:
        print('technology', end = ' ')
        j += 1

    i += 1  #increment the counter
    print()
```

```python
i = 1

while i <= 5:  #condition
    print('data science :', end = ' * ')
    j = 1
    while j <= 4:
        print('technology', end = ' * ')
        j += 1

    i += 1  #increment the counter
    print()
```

```python
i = 1

while i <= 2:  #condition
    j = 0
    while j <= 2:
        print(i*j, end = ' ')
        j += 1
    print()
    i += 1  #increment the counter
    
```

```python
i = 1
while i <= 4:
    j = 0
    while j <= 3:
        print(i*j, end = ' ')
        j += 1
    print()
    i += 1  #increment the counter
```
