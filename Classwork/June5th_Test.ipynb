```python
count = 0
for i in range(1, 21):
    if i % 2 == 0:
        continue
    if i % 7 == 0:
        break
    count += 1
print(count)
```

```python
for num in range(1, 6):
    match num % 2:
        case 0:
            print(f"{num} is even", end='; ') 
        case 1:
            print(f"{num} is odd", end='; ')
        case _:
            print(f"{num} is something else", end='; ')
```
