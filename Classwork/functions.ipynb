VARIABLE LENGTH ARGUMENT

```python
def sum(a,b):
    c = a+b
    print(c)

sum(5,6,7,8)
```

```python
def sum(a,*b):
    c = a+b
    print(c)

sum(5,6,7,8)
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))
    print(a)
    print(b)

sum(5,6,7,8)
```

```python
def sum(a,*b):
    c = a
    for i in b:
        c = c+i
    print(c)

sum(5,6,7,8)
```

```python
def sum(a,*b):
    c = a
    for i in b:
        c = c+i
    print(c)

sum(5,6,7,8,9,10,-1,-2,100)
```

```python
def sum(a,*b):
    c = 0
    for i in b:
        c = c+i
    print(c)

sum(5,6,7,8)
```

KWARGS (KW+ARGS)(KEYWORDS+ARGUMENTS)

```python
def person():
    person('ALEX', 36, 'JOHN', 987767)
```

```python
def person(name, *data):
    print(name)
    print(data)
person('ALEX', 36, 'JOHN', 987767)
```

```python
def person(name, *data):
    print('name')
    print(data)
    
person('ALEX', age = 36, home_place='southcity', mob= 987767)
```

```python
def person(name, **data):
    print('name')
    print(data)
    
person('ALEX', age = 36, home_place='southcity', mob= 987767)
```
